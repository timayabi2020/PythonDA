```python
string = input('Enter a string')
print(string[::-1])
```

```python
# using for loop
string = input('Enter a string')
reversed_text = ""
for char in string:
    reversed_text = char + reversed_text

print(reversed_text)
```

```python
text = input('Enter a string')
for char in reversed(text):
    print(char, end=" ")
```

```python
text = input('Enter a string')
for i in range(len(text)-1,-1,-1):
    print(text[i], end =" ")
```
