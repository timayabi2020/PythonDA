```python
s = input('Enter a string')
l = []
for i in s:
    if i not in ('a','e','i','o','u'):
        l.append(i)
print("".join(l))
```

```python
import re
print(re.sub("[aeiou]", "", input("Enter a string")))
```

```python
s = input('Enter a string')
vowel = 0
cons = 0
for i in s:
    if i not in ('a','e','i','o','u') and i.isalpha():
        cons+=1
    else:
        vowel+=1

print(f"Vowels = {vowel}, consonants = {cons}")
```

```python
# Using list comprehension

s = input('Enter a string')
count_letters = lambda x: (
    len([v for v in x if v.lower() in "aeiou"]),
    len([v for v in x if v.lower() not in "aeiou" and v.isalpha()])
)
num = count_letters(s)
print(f"No of vowels = {num[0]}, No of consonants = {num[1]}")
```
