```python
num = int(input("Enter a number"))
counter = 0
while num != 0:
    counter+=1
    num = num // 10

print(counter)
```

```python
num = int(input("Enter a number"))
search = int(input("Enter digit to search"))
counter = 0
while num != 0:
    digit = num % 10
    if digit == search:
        counter+=1
    num = num // 10

print(f"{search} occurs {counter} times")
```
