```python
num = input('Enter a number')

sum = 0

for digit in num:
    sum+=int(digit)

print(sum)
```

```python
num = input('Enter a number')
counter = len(num)
total = 0

while counter > 0:
    total+=int(num[counter-1])
    counter-=1

print(total)
```

```python
# list comprehension
num = input('Enter a number')
print(sum([int(x) for x in num]))
```

```python
# anonymous function

num = input("Enter a number: ")

tot = lambda x: sum(int(digit) for digit in x)

print(tot(num))
```

```python
# using a map

num = input("Enter a number: ")

tot = lambda x: sum(map(int, x))

print(tot(num))
```

```python
# Sum of Odd digits and sum of even digits

num = input("Enter a number")

sum_of_odd = sum([int(x) for x in num if int(x) % 2 !=0])
sum_of_even = sum([int(x) for x in num if int(x) % 2 ==0])

print(sum_of_odd)
print(sum_of_even)
```

```python
# Sum of Odd digits and sum of even digits.Anonymous functiosn

num = input("Enter a number")

sum_of_odd = lambda x: sum(int(digits) for digits in x if int(digits) % 2 != 0)

print(sum_of_odd(num))
```

```python
# One linere
num = input("Enter a number")
sum_odd_even = lambda x: (sum(d for d in map(int, x) if d % 2 != 0), sum(d for d in map(int, x) if d % 2 == 0))

total_sum = sum_odd_even(num)

print('Sum of odd ', total_sum[0])
print('Sum of even ', total_sum[1])
```
