```python
num = int(input("Enter a number"))
n1, n2 = 0, 1
print(n1, n2, end=" ")
for i in range(2, num):
    n3 = n1 + n2
    print(n3, end=" ")
    n1 = n2
    n2 = n3
```

```python
# using recursion

def fibonacci(n):
    if n == 0 or n == 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

num = int(input("Enter a number"))
for i in range(0,num):
    print(fibonacci(i), end=" ")
```

```python
#Special number
num = int(input("Enter a number"))
result = 0
while num!= 0:
    digit1 = num % 10
    digit2 = num // 10
    result = (digit1+digit2)+(digit1*digit2)
    print("Special" if result==num else "not special")
    num = num // 10
```

```python
#Print all the special 2 digit numbers from 10-99

for num in range(10, 100):
    i = num //10
    j = num % 10

    if (i + j) + (i * j) == num:
        print(num)
```
