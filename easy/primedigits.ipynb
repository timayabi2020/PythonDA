```python
num = int(input('Enter number'))

result = 0

while num != 0:
    digits = num % 10
    if digits in (2,3,5,7):
        result+=digits
    num = num // 10

print(result)
```

```python
num = input('Enter number')

result = lambda x: sum(int(digits) for digits in x if int(digits) in (2,3,5,7))

print(result(num))
```

```python
num = input('Enter number')

result = lambda x: sum(int(digits) for digits in x if int(digits) % 3 == 0)

print(result(num))
```
