```python
num = input("Enter a number")
print('Palindrome ' if num==num[::-1] else 'Not a palindrome')
```

```python
text = input("Enter string")
print("Palindrome " if text == text[::-1] else "Not a palindrome ")
```

```python
# Using for loop

text = input("Enter string")
pld = ""

for s in range(len(text)-1,-1,-1):
    pld+=text[s]

print('Palindrome ' if text==pld else 'Not a palindrome')
```

```python
text = input("Enter string")
reverse_string = lambda x: x[::-1]
print('Palindrome ' if text==reverse_string(text) else 'Not a palindrome')
```
