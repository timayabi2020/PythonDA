```python
num = int(input("Enter a number"))
n1, n2 = 0, 1
print(n1, n2, end=" ")
for i in range(2, num):
    n3 = n1 + n2
    print(n3, end=" ")
    n1 = n2
    n2 = n3
```

```python
# using recursion

def fibonacci(n):
    if n == 0 or n == 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

num = int(input("Enter a number"))
for i in range(0,num):
    print(fibonacci(i), end=" ")
```
